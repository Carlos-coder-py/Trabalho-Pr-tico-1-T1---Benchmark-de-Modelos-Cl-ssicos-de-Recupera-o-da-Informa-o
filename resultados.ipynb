# Resultados — Benchmark de modelos clássicos de RI

Aqui estão os **resultados do trabalho**:
1. **Escalabilidade** (item 4.1): quanto tempo cada modelo leva conforme o corpus cresce.
2. **Relevância** (item 4.2): o que cada modelo devolve em consultas difíceis.

Os modelos estão no arquivo `app.py` (o mesmo usado na demo do Streamlit).

```python
import time
import random
from collections import Counter
import matplotlib.pyplot as plt
import pandas as pd
from app import carregar_corpus, tokenizar, MODELOS

corpus = carregar_corpus()
textos = corpus['texto'].tolist()
print('Documentos no corpus:', len(textos))
```

## 1. Escalabilidade (latência × tamanho do corpus)

**Como medimos:**
- Tamanhos: 100, 500, 1.000 e o corpus inteiro (4.156). O enunciado sugere até 10.000, mas nosso corpus tem menos que isso, então o máximo é o total.
- 20 consultas iguais para todos os modelos (com 1, 2 e 3 palavras), geradas a partir das palavras do menor subconjunto — assim todas existem em todos os tamanhos.
- Cada consulta é repetida 10 vezes e tiramos a média (em milissegundos).
- O Booleano recebe as palavras unidas por `AND`.

```python
def gerar_consultas(textos_pequenos, quantidade=20, semente=42):
    """Sorteia palavras (que aparecem em pelo menos 2 documentos) e monta consultas de 1, 2 e 3 palavras."""
    docs_por_palavra = Counter(p for t in textos_pequenos for p in set(tokenizar(t)))
    palavras = sorted(p for p, n in docs_por_palavra.items() if n >= 2 and p.isalpha() and len(p) >= 3)
    sorteio = random.Random(semente)
    tamanhos = [1] * 7 + [2] * 7 + [3] * (quantidade - 14)
    return [sorteio.sample(palavras, t) for t in tamanhos]


def medir(modelo, consulta, repeticoes=10):
    modelo.buscar(consulta)                      # 1ª execução só para "aquecer"
    inicio = time.perf_counter()
    for _ in range(repeticoes):
        modelo.buscar(consulta)
    return (time.perf_counter() - inicio) / repeticoes * 1000   # ms


tamanhos = [100, 500, 1000, len(textos)]
consultas = gerar_consultas(textos[:100])
print('Exemplos de consultas:', consultas[:3], consultas[8:10], consultas[16:18])
```

```python
linhas = []
for n in tamanhos:
    for nome, Classe in MODELOS.items():
        inicio = time.perf_counter()
        modelo = Classe().indexar(textos[:n])                     # constrói o índice
        tempo_indice = (time.perf_counter() - inicio) * 1000
        conector = ' AND ' if nome == 'Booleano' else ' '
        tempos = [medir(modelo, conector.join(q)) for q in consultas]
        linhas.append({'N': n, 'modelo': nome,
                       'consulta_ms': sum(tempos) / len(tempos), 'indexacao_ms': tempo_indice})

resultado = pd.DataFrame(linhas)
print('Tempo médio por consulta (ms):')
resultado.pivot(index='N', columns='modelo', values='consulta_ms')[list(MODELOS)].round(3)
```

```python
print('Tempo para construir o índice (ms):')
resultado.pivot(index='N', columns='modelo', values='indexacao_ms')[list(MODELOS)].round(1)
```

```python
fig, eixos = plt.subplots(1, 2, figsize=(13, 4.5))
for eixo, escala in zip(eixos, ['linear', 'log']):
    for nome, g in resultado.groupby('modelo', sort=False):
        eixo.plot(g['N'], g['consulta_ms'], marker='o', label=nome)
    eixo.set_yscale(escala)
    eixo.set_title(f'Latência por consulta (escala {escala})')
    eixo.set_xlabel('Tamanho do corpus (N)'); eixo.set_ylabel('ms'); eixo.grid(alpha=.3)
eixos[0].legend()
plt.tight_layout(); plt.show()
```

```python
lin = resultado[resultado.modelo == 'Linear'].set_index('N')['consulta_ms']
print(f"Corpus {tamanhos[-1] // tamanhos[0]}x maior -> Linear ficou {lin[tamanhos[-1]] / lin[tamanhos[0]]:.0f}x mais lento")
for nome in ['BM25', 'Booleano', 'TF-IDF', 'LSA']:
    s = resultado[resultado.modelo == nome].set_index('N')['consulta_ms']
    print(f"{nome}: ficou {s[tamanhos[-1]] / s[tamanhos[0]]:.1f}x mais lento")
```

### O que os gráficos mostram

1. **Linear cresce junto com o corpus** (O(N)): corpus ~40x maior, tempo ~40x maior. É o esperado, pois varre todos os textos a cada busca. Com poucos documentos é rápido, mas ultrapassa o TF-IDF perto de 500 documentos e o LSA perto de 1.000.
2. **Booleano e BM25 crescem muito pouco**: só mexem nos dados das palavras da consulta, não no corpus todo. O Booleano é o mais rápido (só operações de bit).
3. **TF-IDF e LSA têm um custo fixo maior** (multiplicação de matrizes), mas crescem devagar. O LSA é o mais caro por consulta.
4. **Indexação:** o Linear praticamente não tem (só guarda os textos). Os demais pagam uma vez para montar o índice; o LSA paga mais (o SVD).
5. **Conclusão:** índice invertido/matriz vale a pena quando há muitas consultas. Aqui tudo é rápido (poucos ms) porque o corpus é pequeno; a diferença viraria problema com milhões de documentos.

*Os valores exatos mudam de computador para computador; o formato das curvas é o que importa.*

## 2. Relevância — consultas difíceis (Top-5)

Escolhemos 5 consultas, cada uma testando uma dificuldade (identificadas na exploração):

| Consulta | Dificuldade |
|---|---|
| `dress` | termo muito frequente |
| `cotton kurta` | duas palavras (material + peça) |
| `perfume` | sinônimo (o catálogo usa `edp`, `fragrance`) |
| `set` | polissemia (kurta set, bracelet set, gift set...) |
| `shrug` | termo raro |

Comparamos TF-IDF, BM25 e LSA. Também mostramos quantos documentos o Linear e o Booleano acharam.

```python
modelos = {nome: Classe().indexar(textos) for nome, Classe in MODELOS.items()}

def comparar(consulta, k=5):
    print(f'Consulta: "{consulta}"')
    print('  Linear (substring) achou:', len(modelos['Linear'].buscar(consulta, k=len(textos))), 'documentos')
    print('  Booleano (AND) achou:   ', len(modelos['Booleano'].documentos_que_atendem(consulta)), 'documentos')
    tabela = {}
    for nome in ['TF-IDF', 'BM25', 'LSA']:
        tabela[nome] = [f"{corpus.loc[i, 'descricao'][:48]} ({s:.2f})" for i, s in modelos[nome].buscar(consulta, k)]
        tabela[nome] += [''] * (k - len(tabela[nome]))
    pd.set_option('display.max_colwidth', 70)
    return pd.DataFrame(tabela, index=range(1, k + 1))
```

```python
comparar('dress')
```

**`dress` (termo frequente):** os três modelos devolvem vestidos, então "acertam", mas os **5 primeiros são todos diferentes** entre os modelos. Com centenas de vestidos igualmente bons, o desempate é quase arbitrário. No BM25 os 5 primeiros têm exatamente o mesmo score (a frequência satura e os documentos têm tamanho parecido). Além disso, o **Linear achou mais documentos que o Booleano** porque busca substring (pega "dresses", por exemplo): falsos positivos.

```python
comparar('cotton kurta')
```

**`cotton kurta` (duas palavras):** todos trazem kurtas de algodão. O **Booleano (AND) exige as duas palavras** e devolve dezenas/centenas de documentos sem ordem. O **Linear** conta ocorrências de qualquer uma das palavras e por isso devolve muito mais documentos (todos com `cotton` *ou* `kurta`), misturando coisas menos relevantes.

```python
comparar('perfume')
```

**`perfume` (sinônimo):** só poucos produtos têm a palavra "perfume". **TF-IDF e BM25 trazem só esses.** O **LSA traz outros perfumes "EDP" (eau de parfum) que nem contêm a palavra** "perfume" — ele aprendeu que `edp` aparece no mesmo contexto. Esse é o ponto forte do LSA. Contrapartida: o LSA não garante que a palavra buscada esteja no resultado.

```python
comparar('set')
```

**`set` (polissemia):** a palavra tem vários sentidos no catálogo. **TF-IDF e BM25** privilegiam joias ("bracelet set", "set of 4"), onde a palavra aparece mais. O **LSA mistura sentidos** (prendedores de cabelo, conjuntos de roupa). Nenhum modelo *entende* o sentido; o LSA só mostra vários contextos ao mesmo tempo.

```python
comparar('shrug')
```

**`shrug` (termo raro):** só 2 produtos têm a palavra. **TF-IDF e BM25 devolvem exatamente esses 2**: palavra rara tem IDF alto, então não há dúvida. O **LSA coloca esses 2 primeiro, mas completa o Top-5 com produtos só parecidos** (kurta, moletom, blusa), com scores mais baixos. Bom para achar mais coisas (recall), pior em precisão.

## 3. Conclusão

| Modelo | Quando usar | Limitação |
|---|---|---|
| **Linear** | baseline / corpus minúsculo | tempo cresce com o corpus; falsos positivos por substring |
| **Booleano** | filtros exatos (AND, OR, NOT) | sem ranking; AND demais pode dar zero resultados |
| **TF-IDF** | busca simples com ranking | não controla saturação nem tamanho do documento |
| **BM25** | **melhor equilíbrio velocidade × precisão** para busca por palavras | precisa das palavras exatas; empata em termos comuns |
| **LSA** | sinônimos e busca exploratória | mais lento; resultados vagos; não resolve polissemia |

**Limitação do estudo:** a relevância foi avaliada olhando o Top-5, sem julgamento humano formal (precision@k, nDCG). Seria o próximo passo.
